# 2. Differential Equations

Now we connect this directly to physical systems.

A differential equation describes a relationship between a quantity and its derivatives.

For example:

$$\dfrac{dx}{dt} = ax$$

This equation says:

The rate of change of $x$ is proportional to $x$.

We can solve it.

Separate variables:

$$\dfrac{1}{x} dx = a dt$$

Integrate:

$$ ln|x| = at + C$$

Therefore:

$$x(t) = C e^{a t}$$

And look what happened.

The solution of a differential equation naturally produced an exponential.

That's why exponentials and complex numbers are so deeply connected to control.

## First-order differential equation

Consider:

$$ \dot{x} + ax = 0 $$


where:

$$\dot{x} = \dfrac{dx}{dt}$$

Assume:

$$x = e^{st}$$

Then:

$$\dot{x} = s e^{st}$$

Substituting:

$$ s e^{st} + a e^{st} = 0$$

Factor:

$$ e^{st} (s + a) = 0$$

Since:

$$e^{st} \neq 0$$

we get:

$$s+a=0$$

Therefore:

$$s=−a$$
	​


This $s$ is essentially the natural mode of the system.

And we're already approaching the concept of $a$ pole.

## Example: RC circuit

Consider a capacitor discharging through a resistor, where:

$$ \sum V = 0$$

$$ v_c(t) + v_r (t)= 0 $$

Remember that:

$$ q = C v_c(t)$$

$$ v_R(t) = R I(t) $$

$$ I(t) = \dfrac{dq}{dt}$$

Therefore:

$$
 \dfrac{d v_c(t)}{dt} = \dfrac{1}{C} \dfrac{dq}{dt} \rightarrow \dfrac{d v_c(t)}{dt} = \dfrac{1}{C} I(t) 
 \rightarrow  I(t) = C \dfrac{d v_c(t)}{dt}
$$ 

Substituting: 

$$ v_c(t) + RC \dfrac{d v_c(t)}{dt}= 0 $$

The governing equation is:

$$RC \dfrac{dv_c}{dt} + v_c = 0$$

Divide by RC:

$$\dfrac{dv_c}{dt} + \dfrac{1}{RC} v_c = 0$$

Comparing with:

$$\dot{x} + ax = 0$$

we have:

$$a = \dfrac{1}{RC}$$

Therefore:

$$v_c = V_0 e^{-t/(RC)}$$

The quantity

$$τ=RC$$

is the time constant.

And now you can already see a direct relationship:

$$ s = -\dfrac{1}{RC}$$
	​


So the physical parameters $R$ and $C$ determine the mathematical location of the system's pole.

This is exactly the bridge between physics and control theory that we want to develop.

```python
import numpy as np
import matplotlib.pyplot as plt

# Circuit parameters
R = 1000          # Resistance [Ohm]
C = 1000e-6       # Capacitance [F]
V0 = 5             # Initial voltage [V]

# Time constant
tau = R * C

# Time
t = np.linspace(0, 5 * tau, 1000)

# Capacitor voltage during discharge
Vc = V0 * np.exp(-t / tau)

# Plot
plt.figure(figsize=(10, 5))

plt.plot(t, Vc, linewidth=2)

# Time constant marker
plt.axvline(tau, linestyle="--", linewidth=1.5)

# Labels
plt.xlabel("Time [s]")
plt.ylabel("Capacitor Voltage [V]")
plt.title(
    f"RC Circuit Discharge\n"
    f"R = {R} Ω, C = {C*1e6:.0f} μF, τ = RC = {tau:.2f} s"
)

plt.grid(True)

# Show values
plt.text(
    tau,
    V0 * np.exp(-1),
    f"  τ = {tau:.2f} s\n  V(τ) = {V0*np.exp(-1):.2f} V",
    verticalalignment="bottom"
)

plt.show()
```
