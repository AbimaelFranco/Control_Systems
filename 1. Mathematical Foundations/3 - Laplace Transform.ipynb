# 3. Laplace Transform

The central idea of this chapter is:

$$ \text{Differential equation in time} \rightarrow \text{Algebraic equation in s}$$

And eventually:

$$ \text{Physical system} \rightarrow \text{Differential equation} \rightarrow \text{Laplace transform} \rightarrow \text{Transfer function} \rightarrow \text{Control analysis} $$

## Why do we need the Laplace transform?

Suppose we have a simple RC circuit:

$$ RC \dfrac{dv_c(t)}{dt} + v_c(t) = v_{in}(t)$$

Trying to solve this directly in the time domain is possible, but things become increasingly difficult when we have:

* higher-order systems,
* multiple differential equations,
* discontinuous inputs,
* initial conditions,
* feedback,
* complicated forcing functions.

The Laplace transform gives us a different representation.

Instead of working with:

$$\dfrac{dv}{dt}$$

we can work with:

$$ sV(s) - v(0^-)$$

And instead of solving a differential equation, we can manipulate an algebraic equation.

That is enormously useful in control.

## The basic definition

The unilateral Laplace transform is defined as:

$$ \mathcal{L} \{f(t) \} = F(s) = \int_0^\infty f(t) e^{-st} dt$$

where:

$$ s=\sigma +j \omega $$

Notice something important:

The s from Laplace is the same complex variable we were discussing in the previous chapter.

We are not introducing an unrelated variable.

## Why is s complex?

Remember:

$$ s= \sigma +j \omega $$

Therefore:

$$ e^{-st} = e^{-(\sigma + j\omega) t}$$

which can be separated:

$$ e^{-st} = e^{- \sigma t} e^{- j\omega t}$$


Using Euler:

$$ e^{- j\omega t} = \cos(\omega t) + j \sin(\omega t) $$

Therefore:

$$ e^{-st} = e^{- \sigma t} \bigg[ \cos(\omega t) + j \sin(\omega t) \bigg]$$

The Laplace transform is essentially analyzing a signal through a combination of:

* exponential decay/growth
* oscillation

This is one reason complex numbers are so fundamental to control.

## Foundamentals Laplace transforms

### Constant transform

Let's calculate:

$$ \mathcal{L} \{ 1 \} $$

By definition:

$$ \mathcal{L} \{ 1 \} = \int_0^\infty e^{-st} dt$$

Integrating:

$$ =\bigg[ -\dfrac{1}{s} e^{-st}\bigg]_0^\infty$$

For:

$$Re(s)>0$$

the exponential goes to zero:

$$e^{-st} \rightarrow 0$$

Therefore:

$$ \mathcal{L} \{ 1 \} = \dfrac{1}{s}$$

This result is extremely important.

A constant signal in the time domain becomes:

$$\dfrac{1}{s}$$

in the s-domain.

### The unit step

The unit step is:

$$
u(t) =
\begin{cases}
0, & t < 0 \\
1, & t \geq 0
\end{cases}
$$

For $t\geq 0$, it is simply $1$.

Therefore:

$$ \mathcal{L} \{ u(t) \} = \dfrac{1}{s}$$

This is why you'll constantly see:	​

$$\dfrac{1}{s}$$

when analyzing step responses.

### Exponential functions

Consider:

$$f(t) = e^{a t}$$

Then:

$$ \mathcal{L} \{ e^{at} \} = \int_0^\infty e^{at} e^{-st} dt$$

Combine exponents:

$$ = \int_0^\infty e^{-(s-a)t}dt$$

Therefore:

$$ \mathcal{L} \{ e^{at} \} = \dfrac{1}{s-a} $$


This is another fundamental result.

For example:

$$e^{-3t}$$

becomes:

$$\dfrac{1}{s+3}$$

Notice the relationship:

$$ e^{-3t} \leftrightarrow \dfrac{1}{s+3}$$

The exponential decay rate appears directly in the denominator.

### Sinusoids

For:

$$f(t)=\sin(\omega t)$$

we obtain:

$$ \mathcal{L} \{ \sin(\omega t) \} = \dfrac{\omega}{s^2 + \omega^2} $$

Similarly, changing the numerator:

$$ \mathcal{L} \{ \cos(\omega t) \} = \dfrac{s}{s^2 + \omega^2} $$

These are important because oscillatory systems naturally generate sine and cosine terms.

### The derivative property

Now we reach the most important property for differential equations.

If:

$$ \mathcal{L} \{ f(t) \} = F(s)$$

Then:

$$ \mathcal{L} \bigg\{ \dfrac{df}{dt} \bigg\} = sF(s) - f(0^-)$$

This is the reason Laplace transforms are so useful.

A derivative becomes an algebraic multiplication by s.

Ignoring initial conditions for a moment:

$$ \dfrac{df}{dt} \rightarrow sF(s)$$

and:

$$ \dfrac{d^2f}{dt^2} \rightarrow s^2F(s)$$

This means:

$$ \dfrac{d}{dt} \rightarrow s$$


under zero initial conditions.

That's not literally an equality between operators in every context, but it is an incredibly useful engineering interpretation.

But, where's $-f(0^-)$ term comming? Starting from:

$$ \mathcal{L} \{ f'(t) \} = \int_0^\infty f'(t) e^{-st} dt $$

Use integration by parts:

$$ u = e^{-st} $$

and:

$$ dv = f'(t) dt $$

Then:

$$ du = -se^{-st} dt$$

and:

$$v = f(t)$$

Therefore:

$$ \int_0^\infty f'(t) e^{-st} dt = \big[f(t) e^{-st}\big]_0^\infty + s \int_0^\infty f(t) e^{-st} dt$$

The boundary term becomes:

$$-f(0^-)$$

so:

$$ \mathcal{L} \bigg\{ \dfrac{df}{dt} \bigg\} = sF(s) - f(0^-)$$

This derivation is worth understanding rather than memorizing.

Other case is higher-order derivates. Apply the derivative property again:

$$ \mathcal{L} \bigg\{ f''(t) \bigg\} = s \mathcal{L} \{f'(t)\} - f'(0^-)$$

Therefore:

$$ \mathcal{L} \bigg\{ f''(t) \bigg\} = s^2F(s) - sf(0^-) - f(0^-)$$

For zero initial conditions:

$$ f(0)=0 $$

and:

$$ f'(0) = 0 $$

so:

$$ \mathcal{L} \bigg\{ f''(t) \bigg\} = s^2F(s) $$

And similarly:

$$ \mathcal{L} \bigg\{ f^{(n)}(t) \bigg\} = s^nF(s) $$

under zero initial conditions.

### The integration property

The integration property is essentially the counterpart to the derivative property.

Suppose:

$$ F(s)=\mathcal{L} \{ f(t) \} $$

and define

$$ g(t) = \int_0^t f(\tau) d\tau$$

Then:

$$ \mathcal{L} \bigg\{ \int_0^t f(\tau) d\tau \bigg\} = \dfrac{F(s)}{s}$$

assuming the integral starts at $t=0$.

So, under zero initial conditions:

$$\dfrac{d}{ds} \leftrightarrow s $$

and

$$ \int dt \leftrightarrow \dfrac{1}{s} $$

This is an extremely useful mental model.

## Example

Transforming a differential equation

Now let's see why all of this matters.

Consider:

$$ \ddot{x} + 3\dot{x} + 2x = u(t)$$

Assume zero initial conditions.

Taking the Laplace transform:

$$ s^2 X(s) + 3sX(s) + 2 X(s) = \dfrac{1}{s} $$

Factor:

$$ X(s) (s^2 + 3s + 2) = \dfrac{1}{s} $$

Therefore:

$$X(s) = \dfrac{1}{s(s^2 + 3s + 2)}$$

Factor the polynomial:

$$ s^2 + 3s + 2 = (s+1) (s+2)$$

so:

$$X(s) = \dfrac{1}{s(s+1)(s+2)}$$

We've transformed a differential equation into an algebraic expression.
