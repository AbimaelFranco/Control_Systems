# 3. Laplace Transform

The central idea of this chapter is:

$$ \text{Differential equation in time} \rightarrow \text{Algebraic equation in s}$$

And eventually:

$$ \text{Physical system} \rightarrow \text{Differential equation} \rightarrow \text{Laplace transform} \rightarrow \text{Transfer function} \rightarrow \text{Control analysis} $$

## Why do we need the Laplace transform?

Suppose we have a simple RC circuit:

$$ RC \dfrac{dv_c(t)}{dt} + v_c(t) = v_{in}(t)$$

Trying to solve this directly in the time domain is possible, but things become increasingly difficult when we have:

* higher-order systems,
* multiple differential equations,
* discontinuous inputs,
* initial conditions,
* feedback,
* complicated forcing functions.

The Laplace transform gives us a different representation.

Instead of working with:

$$\dfrac{dv}{dt}$$

we can work with:

$$ sV(s) - v(0^-)$$

And instead of solving a differential equation, we can manipulate an algebraic equation.

That is enormously useful in control.

## The basic definition

The unilateral Laplace transform is defined as:

$$ \mathcal{L} \{f(t) \} = F(s) = \int_0^\infty f(t) e^{-st} dt$$

where:

$$ s=\sigma +j \omega $$

Notice something important:

The s from Laplace is the same complex variable we were discussing in the previous chapter.

We are not introducing an unrelated variable.

## Why is s complex?

Remember:

$$ s= \sigma +j \omega $$

Therefore:

$$ e^{-st} = e^{-(\sigma + j\omega) t}$$

which can be separated:

$$ e^{-st} = e^{- \sigma t} e^{- j\omega t}$$


Using Euler:

$$ e^{- j\omega t} = \cos(\omega t) + j \sin(\omega t) $$

Therefore:

$$ e^{-st} = e^{- \sigma t} \bigg[ \cos(\omega t) + j \sin(\omega t) \bigg]$$

The Laplace transform is essentially analyzing a signal through a combination of:

* exponential decay/growth
* oscillation

This is one reason complex numbers are so fundamental to control.

## Foundamentals Laplace transforms

### Constant transform

Let's calculate:

$$ \mathcal{L} \{ 1 \} $$

By definition:

$$ \mathcal{L} \{ 1 \} = \int_0^\infty e^{-st} dt$$

Integrating:

$$ =\bigg[ -\dfrac{1}{s} e^{-st}\bigg]_0^\infty$$

For:

$$Re(s)>0$$

the exponential goes to zero:

$$e^{-st} \rightarrow 0$$

Therefore:

$$ \mathcal{L} \{ 1 \} = \dfrac{1}{s}$$

This result is extremely important.

A constant signal in the time domain becomes:

$$\dfrac{1}{s}$$

in the s-domain.

### The unit step

The unit step is:

$$
u(t) =
\begin{cases}
0, & t < 0 \\
1, & t \geq 0
\end{cases}
$$

For $t\geq 0$, it is simply $1$.

Therefore:

$$ \mathcal{L} \{ u(t) \} = \dfrac{1}{s}$$

This is why you'll constantly see:	​

$$\dfrac{1}{s}$$

when analyzing step responses.

### Exponential functions

Consider:

$$f(t) = e^{a t}$$

Then:

$$ \mathcal{L} \{ e^{at} \} = \int_0^\infty e^{at} e^{-st} dt$$

Combine exponents:

$$ = \int_0^\infty e^{-(s-a)t}dt$$

Therefore:

$$ \mathcal{L} \{ e^{at} \} = \dfrac{1}{s-a} $$


This is another fundamental result.

For example:

$$e^{-3t}$$

becomes:

$$\dfrac{1}{s+3}$$

Notice the relationship:

$$ e^{-3t} \leftrightarrow \dfrac{1}{s+3}$$

The exponential decay rate appears directly in the denominator.

### Sinusoids

For:

$$f(t)=\sin(\omega t)$$

we obtain:

$$ \mathcal{L} \{ \sin(\omega t) \} = \dfrac{\omega}{s^2 + \omega^2} $$

Similarly, changing the numerator:

$$ \mathcal{L} \{ \cos(\omega t) \} = \dfrac{s}{s^2 + \omega^2} $$

These are important because oscillatory systems naturally generate sine and cosine terms.
